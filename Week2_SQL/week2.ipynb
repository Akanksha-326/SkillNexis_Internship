### 1. Environment Setup & Database Connection

```python
import sqlite3
import pandas as pd

# Connect to sales.db
conn = sqlite3.connect('sales.db')

# Helper function to run and format SQL queries
def run_query(sql):
    return pd.read_sql_query(sql, conn)

print("Connected to sales.db successfully!")
```

### 2. Basic Queries (`SELECT`, `WHERE`, `ORDER BY`, `LIMIT`)

```python
# Retrieve the first 10 sales records
query = """
SELECT order_id, customer_name, order_date, category, product_name, quantity, unit_price, total_price, region
FROM sales
LIMIT 10;
"""
run_query(query)
```

```python
# Filter records for the 'West' region
query = """
SELECT order_id, customer_name, product_name, total_price, region
FROM sales
WHERE region = 'West'
LIMIT 10;
"""
run_query(query)
```

```python
# Filter records with multiple conditions: Electronics with quantity >= 5
query = """
SELECT order_id, customer_name, category, product_name, quantity, total_price
FROM sales
WHERE category = 'Electronics' AND quantity >= 5;
"""
run_query(query)
```

```python
# Sort sales from highest to lowest total price
query = """
SELECT order_id, customer_name, category, total_price
FROM sales
ORDER BY total_price DESC
LIMIT 10;
"""
run_query(query)
```

### 3. Aggregate Functions (`SUM`, `AVG`, `COUNT`, `MIN`, `MAX`)

```python
# Key summary metrics across all sales orders
query = """
SELECT 
    COUNT(*) AS total_orders,
    SUM(total_price) AS total_revenue,
    ROUND(AVG(total_price), 2) AS average_order_value,
    MIN(total_price) AS min_order_value,
    MAX(total_price) AS max_order_value
FROM sales;
"""
run_query(query)
```

```python
# Total units sold and unique customer/category counts
query = """
SELECT 
    SUM(quantity) AS total_units_sold,
    COUNT(DISTINCT customer_name) AS unique_customers,
    COUNT(DISTINCT category) AS unique_categories
FROM sales;
"""
run_query(query)
```

### 4. Grouping & Filtering (`GROUP BY`, `HAVING`)

```python
# Revenue and volume summarized by product category
query = """
SELECT 
    category,
    COUNT(*) AS total_orders,
    SUM(quantity) AS total_units_sold,
    SUM(total_price) AS total_revenue,
    ROUND(AVG(total_price), 2) AS avg_order_value
FROM sales
GROUP BY category
ORDER BY total_revenue DESC;
"""
run_query(query)
```

```python
# Sales distribution by geographic region
query = """
SELECT 
    region,
    COUNT(*) AS total_orders,
    SUM(total_price) AS total_revenue,
    ROUND(AVG(total_price), 2) AS average_order_value
FROM sales
GROUP BY region
ORDER BY total_revenue DESC;
"""
run_query(query)
```

```python
# Categories with total revenue exceeding 500,000 using HAVING
query = """
SELECT 
    category,
    COUNT(*) AS order_count,
    SUM(total_price) AS total_revenue
FROM sales
GROUP BY category
HAVING SUM(total_price) > 500000
ORDER BY total_revenue DESC;
"""
run_query(query)
```

### 5. Relational Joins (`INNER JOIN`, `LEFT JOIN`)

```python
# INNER JOIN between Orders and Customers
query = """
SELECT 
    o.order_id,
    c.customer_name,
    c.region,
    o.order_date,
    o.total_price
FROM orders o
INNER JOIN customers c 
    ON o.customer_id = c.customer_id
ORDER BY o.total_price DESC
LIMIT 10;
"""
run_query(query)
```

```python
# Multi-Table INNER JOIN (Orders + Customers + Products)
query = """
SELECT 
    o.order_id,
    c.customer_name,
    c.region,
    p.category,
    p.product_name,
    o.quantity,
    p.unit_price,
    o.total_price
FROM orders o
INNER JOIN customers c 
    ON o.customer_id = c.customer_id
INNER JOIN products p 
    ON o.product_id = p.product_id
ORDER BY o.order_id ASC
LIMIT 10;
"""
run_query(query)
```

```python
# LEFT JOIN Regional Targets with Actual Sales & Variance
query = """
SELECT 
    t.region,
    t.regional_manager,
    t.target_revenue,
    COALESCE(SUM(s.total_price), 0) AS actual_revenue,
    ROUND(COALESCE(SUM(s.total_price), 0) - t.target_revenue, 2) AS revenue_variance,
    CASE 
        WHEN COALESCE(SUM(s.total_price), 0) >= t.target_revenue THEN 'Target Achieved'
        ELSE 'Target Not Achieved'
    END AS performance_status
FROM regional_targets t
LEFT JOIN sales s 
    ON t.region = s.region
GROUP BY t.region, t.regional_manager, t.target_revenue
ORDER BY actual_revenue DESC;
"""
run_query(query)
```

### 6. Subqueries (Scalar, `WHERE IN`, Derived Tables)

```python
# Scalar Subquery: Orders higher than overall average order value
query = """
SELECT 
    order_id,
    customer_name,
    category,
    total_price
FROM sales
WHERE total_price > (
    SELECT AVG(total_price) 
    FROM sales
)
ORDER BY total_price DESC
LIMIT 10;
"""
run_query(query)
```

```python
# Multi-Row Subquery with IN: Orders from regions where AOV exceeds 12,000
query = """
SELECT 
    order_id,
    customer_name,
    region,
    total_price
FROM sales
WHERE region IN (
    SELECT region
    FROM sales
    GROUP BY region
    HAVING AVG(total_price) > 12000
)
ORDER BY total_price DESC
LIMIT 10;
"""
run_query(query)
```

```python
# Derived Table Subquery: Difference from Category Average
query = """
SELECT 
    s.order_id,
    s.customer_name,
    s.category,
    s.total_price,
    ROUND(cat_stats.category_avg, 2) AS category_avg_price,
    ROUND(s.total_price - cat_stats.category_avg, 2) AS difference_from_category_avg
FROM sales s
INNER JOIN (
    SELECT 
        category,
        AVG(total_price) AS category_avg
    FROM sales
    GROUP BY category
) cat_stats 
    ON s.category = cat_stats.category
ORDER BY difference_from_category_avg DESC
LIMIT 10;
"""
run_query(query)
```

### 7. CASE Statements (Conditional Logic & Segmentation)

```python
# Order Value Tiering using CASE
query = """
SELECT 
    order_id,
    customer_name,
    category,
    total_price,
    CASE 
        WHEN total_price >= 20000 THEN 'High Value (>= 20k)'
        WHEN total_price >= 10000 THEN 'Medium Value (10k-20k)'
        ELSE 'Low Value (< 10k)'
    END AS order_value_tier
FROM sales
ORDER BY total_price DESC
LIMIT 10;
"""
run_query(query)
```

```python
# Purchase Size Segmentation by Quantity
query = """
SELECT 
    order_id,
    customer_name,
    product_name,
    quantity,
    CASE 
        WHEN quantity >= 8 THEN 'Bulk Purchase'
        WHEN quantity >= 4 THEN 'Standard Purchase'
        ELSE 'Small Purchase'
    END AS purchase_size
FROM sales
ORDER BY quantity DESC
LIMIT 10;
"""
run_query(query)
```

```python
# Conditional Aggregation (Category Breakdown by Region)
query = """
SELECT 
    region,
    COUNT(*) AS total_orders,
    SUM(CASE WHEN category = 'Electronics' THEN total_price ELSE 0 END) AS electronics_revenue,
    SUM(CASE WHEN category = 'Furniture' THEN total_price ELSE 0 END) AS furniture_revenue,
    SUM(CASE WHEN category = 'Grocery' THEN total_price ELSE 0 END) AS grocery_revenue,
    SUM(CASE WHEN category = 'Clothing' THEN total_price ELSE 0 END) AS clothing_revenue,
    SUM(total_price) AS total_regional_revenue
FROM sales
GROUP BY region
ORDER BY total_regional_revenue DESC;
"""
run_query(query)
```

### 8. Assignments: Top Customers & Average Order Values

```python
# Assignment 1.1: Top 10 Customers by Total Spending
query = """
SELECT 
    customer_name,
    region,
    COUNT(*) AS number_of_orders,
    SUM(quantity) AS total_units_purchased,
    SUM(total_price) AS total_spending,
    ROUND(AVG(total_price), 2) AS average_order_value
FROM sales
GROUP BY customer_name, region
ORDER BY total_spending DESC
LIMIT 10;
"""
run_query(query)
```

```python
# Assignment 1.2: Top Customer per Geographic Region
query = """
SELECT 
    s1.region,
    s1.customer_name,
    s1.total_price AS highest_single_order_value
FROM sales s1
WHERE s1.total_price = (
    SELECT MAX(s2.total_price)
    FROM sales s2
    WHERE s2.region = s1.region
)
ORDER BY s1.total_price DESC;
"""
run_query(query)
```

```python
# Assignment 2.1: Overall Store Average Order Value (AOV)
query = """
SELECT 
    COUNT(*) AS total_orders,
    SUM(total_price) AS total_revenue,
    ROUND(AVG(total_price), 2) AS overall_average_order_value
FROM sales;
"""
run_query(query)
```

```python
# Assignment 2.2: Average Order Value (AOV) by Product Category
query = """
SELECT 
    category,
    COUNT(*) AS orders_count,
    SUM(total_price) AS total_revenue,
    ROUND(AVG(total_price), 2) AS average_order_value
FROM sales
GROUP BY category
ORDER BY average_order_value DESC;
"""
run_query(query)
```

```python
# Assignment 2.3: Average Order Value (AOV) by Geographic Region
query = """
SELECT 
    region,
    COUNT(*) AS orders_count,
    SUM(total_price) AS total_revenue,
    ROUND(AVG(total_price), 2) AS average_order_value
FROM sales
GROUP BY region
ORDER BY average_order_value DESC;
"""
run_query(query)
```

```python
# Assignment 2.4: Orders Exceeding Average Order Value
query = """
SELECT 
    order_id,
    customer_name,
    category,
    total_price,
    ROUND((SELECT AVG(total_price) FROM sales), 2) AS overall_avg,
    ROUND(total_price - (SELECT AVG(total_price) FROM sales), 2) AS amount_above_avg,
    ROUND(((total_price - (SELECT AVG(total_price) FROM sales)) / (SELECT AVG(total_price) FROM sales)) * 100, 2) AS pct_above_avg
FROM sales
WHERE total_price > (SELECT AVG(total_price) FROM sales)
ORDER BY total_price DESC
LIMIT 10;
"""
run_query(query)
```
