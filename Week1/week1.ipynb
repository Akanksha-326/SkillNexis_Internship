```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
#Load the Dataset
df = pd.read_excel("global_superstore_2016.xlsx")

print("Dataset loaded successfully!")
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
```

```python
#Display the Dataset
df.head()
```

```python
#Display the Dataset
df.tail()
```

```python
#Column Names
print("Column Names:")
print(df.columns.tolist())
```

```python
#Data Types
print("Data Types:")
print(df.dtypes)
```

```python
#Dataset Information
df.info()
```

```python
#Dataset Size
print("Rows and Columns:", df.shape)
```

```python
#Check Missing Values
missing_values = df.isnull().sum()

print("Missing Values:")
print(missing_values)
```

```python
postal_missing = df["Postal Code"].isnull().sum()
total_rows = len(df)

percentage = (postal_missing / total_rows) * 100

print("Missing Postal Codes:", postal_missing)
print("Total Rows:", total_rows)
print("Missing Percentage:", percentage, "%")
```

```python
df[df["Postal Code"].isnull()][
    ["City", "State", "Country", "Region", "Postal Code"]
].head(20)
```

```python
df["Postal Code"] = df["Postal Code"].fillna("Unknown")
```

```python
print(df["Postal Code"].isnull().sum())
```

```python
print("Missing Values After Cleaning:")
print(df.isnull().sum())
```

```python
#Check Duplicate Values
duplicates = df.duplicated().sum()

print("Number of duplicate rows:", duplicates)
```

```python
#Remove Duplicates
df = df.drop_duplicates()

print("Duplicates removed.")
print("New dataset size:", df.shape)
```

```python
#Convert Order Date
print(df["Order Date"].head())
```

```python
df["Order Date"] = pd.to_datetime(df["Order Date"])

print(df["Order Date"].dtype)
```

```python
#Create Year Column
df["Year"] = df["Order Date"].dt.year

df[["Order Date", "Year"]].head()
```

```python
#Mean
mean_sales = df["Sales"].mean()

print("Mean Sales:", mean_sales)
```

```python
#Median
median_sales = df["Sales"].median()

print("Median Sales:", median_sales)
```

```python
#Mode
mode_sales = df["Sales"].mode()

print("Mode Sales:")
print(mode_sales)
```

```python
#Variance
variance_sales = df["Sales"].var()

print("Variance of Sales:", variance_sales)
```

```python
#Complete Descriptive Statistics
print("DESCRIPTIVE STATISTICS OF SALES")

print("Mean:", df["Sales"].mean())
print("Median:", df["Sales"].median())
print("Mode:", df["Sales"].mode().iloc[0])
print("Variance:", df["Sales"].var())
print("Minimum:", df["Sales"].min())
print("Maximum:", df["Sales"].max())
print("Count:", df["Sales"].count())
```

```python
df["Sales"].describe()
```

```python
#Total Revenue
total_revenue = df["Sales"].sum()

print("Total Revenue:", total_revenue)
```

```python
#Find Categories
print("Categories:")
print(df["Category"].unique())
```

```python
#Sales by Category
category_sales = df.groupby("Category")["Sales"].sum()

print("Sales by Category:")
print(category_sales)
```

```python
#Sort Category Sales
category_sales = df.groupby("Category")["Sales"].sum().sort_values(ascending=False)

print(category_sales)
```

```python
#Category Pivot Table
category_pivot = pd.pivot_table(
    df,
    values="Sales",
    index="Category",
    aggfunc="sum"
)

category_pivot
```

```python
#Yearly Sales
yearly_sales = df.groupby("Year")["Sales"].sum()

print("Yearly Sales:")
print(yearly_sales)
```

```python
#Yearly Sales Pivot Table
yearly_pivot = pd.pivot_table(
    df,
    values="Sales",
    index="Year",
    aggfunc="sum"
)

yearly_pivot
```

```python
#Yearly Sales Graph
plt.figure(figsize=(8, 5))

plt.plot(
    yearly_sales.index,
    yearly_sales.values,
    marker="o"
)

plt.title("Yearly Sales Trend")
plt.xlabel("Year")
plt.ylabel("Total Revenue")

plt.grid()
plt.show()
```

```python
#Sales by Category Graph
plt.figure(figsize=(8, 5))

category_sales.plot(kind="bar")

plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Revenue")

plt.xticks(rotation=0)
plt.show()
```

```python
#DEPARTMENT-WISE REVENUE
print(df.columns.tolist())
```

```python
#Sub-Category-wise Revenue
subcategory_revenue = (
    df.groupby("Sub-Category")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

print("Sub-Category-wise Revenue:")
print(subcategory_revenue)
```

```python
#Sub-Category Pivot Table
subcategory_pivot = pd.pivot_table(
    df,
    values="Sales",
    index="Sub-Category",
    aggfunc="sum"
)

subcategory_pivot = subcategory_pivot.sort_values(
    by="Sales",
    ascending=False
)

subcategory_pivot
```

```python
#Category-Year Pivot Table
category_year_pivot = pd.pivot_table(
    df,
    values="Sales",
    index="Category",
    columns="Year",
    aggfunc="sum"
)

category_year_pivot
```

```python
#Top 10 Sub-Categories
top_10_subcategories = subcategory_revenue.head(10)

print("Top 10 Sub-Categories:")
print(top_10_subcategories)
```

```python
#Top 10 Sub-Categories Graph
plt.figure(figsize=(10, 6))

top_10_subcategories.plot(kind="bar")

plt.title("Top 10 Sub-Categories by Revenue")
plt.xlabel("Sub-Category")
plt.ylabel("Revenue")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
#Sales Analysis
print(" SALES ANALYSIS ")

print("\nTotal Revenue:")
print(df["Sales"].sum())

print("\nMean Sales:")
print(df["Sales"].mean())

print("\nMedian Sales:")
print(df["Sales"].median())

print("\nMode Sales:")
print(df["Sales"].mode().iloc[0])

print("\nVariance:")
print(df["Sales"].var())

print("\nSales by Category:")
print(category_sales)

print("\nYearly Sales:")
print(yearly_sales)

print("\nTop 10 Sub-Categories:")
print(top_10_subcategories)
```
